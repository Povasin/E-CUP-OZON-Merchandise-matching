```python
import sys, os, subprocess, glob
sys.path.insert(0, "/kaggle/input/ecup-matching-src")
# Kaggle распаковывает загруженный архив сам, поэтому точный путь заранее неизвестен —
# ищем рекурсивно, иначе прогон падает после ожидания в очереди.
found = glob.glob("/kaggle/input/**/llm_train.parquet", recursive=True)
if not found:
    raise SystemExit("Пакет не найден в /kaggle/input")
pack = os.path.dirname(found[0])
print("пакет:", pack, os.listdir(pack))
import torch
print("GPU:", torch.cuda.is_available(), torch.cuda.get_device_name(0) if torch.cuda.is_available() else "")
subprocess.run([sys.executable, "-u", "-m", "src.train_ce_large",
                "--prepacked", pack, "--epochs", "1",
                "--batch-size", "256", "--max-length", "192",
                "--output", "/kaggle/working/ce_large"], check=True)
```
